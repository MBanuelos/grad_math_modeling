# Difference Equations <a id='difference'></a>

Difference equations, which are the discrete counterparts to differential equations, have applications in a variety of fields, including Probability, Computer Science, Numerical Methods, and Finance.

Like ODEs, they can be classified according to their **order**, if they **linear** or **non-linear**, and if they are **homogeneous or non-homogeneous**.

### Notation <a id='notation'></a>

We will use $u_0$ as the notation for the first term in a sequence (the *initial condition*) and $u_n$ as the $n$th term in the sequence. We consider time to be discrete and usually $u_n$ is defined with a recurrence relation.

### First Order Equations <a id='first'></a>

**First order homogeneous equations** take the form,

$$
u_n = a u_{n-1}, \; u(0) = u_0.
$$

This is in fact an example of exponential growth or decay!

#### *Ansatz*

In both differential and difference equations, we may start with an initial approach (or guess), called an *ansatz* to show the general form of a solution.

**First order non-homogeneous equations** take the form,

$$
u_n = a u_{n-1} + v_n, \; u(0) = u_0,
$$

where $v_n = v(n)$ is a given sequence. Like in ODEs, there exists a linearity principle for difference equations, and we follow a similar framework to find the general solution:

* Find a particular solution to the non-homogeneous equation, $y_n$
* Find the general solution to the homogeneous equation, $x_n$
* General solution is then $z_n = x_n + y_n$

### Second Order Equations <a id='second'></a>

**Second order homogeneous equations** take the form,

$$
u_n = a u_{n-1} + b u_{n-2}, \; u(0) = u_0,\; u(1) = u_1.
$$

These equations also have their own linearity principle (if $x_n$ and $y_n$ are solutions, then so are their sum). 

Let's use an *ansatz* of $u_n = C \lambda^n$ to begin.

$$
u_n = a u_{n-1} + b u_{n-2} + v_n, \; u(0) = u_0, u(1) = u_1

where $v_n = v(n)$ is a given sequence. As before, we need to find a particular solution and a general one (for homogeneous eq.), and sum them.

# Numerical Methods for Solving ODEs

If we wanted to numerically solve the equation

$$
u''(x) = f(x) \quad \text{0} < x < \text{1}
$$

with some boundary conditions $u(0) = \alpha, u(1) = \beta$.


**Q:** How would you solve this equation?

Since we are solving this problem numerically, then we will solve it over a grid $U_0, U_1, \ldots, U_m, U_{m+1}$, where $U_j$ is our approximation to $u(x_j)$. 

Here, we let $x_j = jh$ and $h = 1/(m+1)$ (mesh width). We know $U_0 = \alpha, U_{m+1} = \beta$. So we need to solve for $U_1, \ldots U_m$.

We can replace $u''(x)$ above by what is known as a *centered difference approximation*

$$
D^2 U_j = \frac{1}{h^2} ( U_{j-1} - 2 U_j + U_{j+1} ),
$$

we obtain

$$
\frac{1}{h^2} ( U_{j-1} - 2 U_j + U_{j+1} = f(x_j) \quad \text{for } j = 1, 2, \ldots, m \Rightarrow AU = F
$$

This is a linear system of $m$ equations with $m$ unknowns and can be readily solved.

**Q:** How do we derive difference approximations for derivatives?

**A:** Taylor series expansions! For example, at a point $\bar{x}$, we have

$$
\begin{align*}
u ( \bar{x} + h ) &= u (\bar{x}) + h u'(\bar{x}) + \frac{1}{2} h^2 u''(\bar{x}) + \frac{1}{6} h^3 u'''(\bar{x}) + \text{higher order terms}\\
u ( \bar{x} - h ) &= u (\bar{x}) - h u'(\bar{x}) + \frac{1}{2} h^2 u''(\bar{x}) - \frac{1}{6} h^3 u'''(\bar{x}) + \text{higher order terms}\\
\end{align*}
$$
