# Subset Selection <a id='subset'></a>


There are 3 main approaches to picking the *best* model:

 * forward stepwise selection
 * backward stepwise selection 
 * best subset selection (exhaustive)


## Best subset selection
**Algorithm 6.1 (ISLR)** *Best subset selection*

1. Let $\mathcal{M}_0$ denote the *null model*, which contains no predictors. This model simply predicts the sample mean for each observation.

2. For $k = 1, 2, \ldots p:$
    * Fit all $p \choose k$ models that contain exactly $k$ predictors.
    * Pick the best model with exactly $k$ predictors, and call it $\mathcal{M}_k$. Here *best* is defined as having smallest RSS (or largest $R^2$).
    
3. Select a single best model from $\mathcal{M}_0, \ldots, \mathcal{M}_p$ using an error metric.

## Forward Selection

**Algorithm 6.2 (ISLR)** *Forward stepwise selection*

1. Let $\mathcal{M}_0$ denote the *null model*, which contains no predictors. 

2. For $k = 0, 1, \ldots p-1:$
    * Consider all $p-k$ models that augment the predictors in $\mathcal{M}_k$ with one additional predictor.
    * Pick the *best* among these $p-k$ models, and call it $\mathcal{M}_{k+1}$. Here *best* is defined as having smallest RSS (or largest $R^2$).
    
3. Select a single best model from $\mathcal{M}_0, \ldots, \mathcal{M}_p$ using an error metric.

## Backward Selection

**Algorithm 6.3 (ISLR)** *Backward stepwise selection*

1. Let $\mathcal{M}_p$ denote the *full model*, which contains all $p$ predictors.

2. For $k = p, p-1, \ldots 1:$
    * Consider all $k$ models that contail all but one of the predictors in $\mathcal{M}_k$, for a total of $k-1$ predictors.
    * Choose the *best* among these $k$ models, and call it $\mathcal{M}_{k-1}$. Here *best* is defined as having smallest RSS (or largest $R^2$).
    
3. Select a single best model from $\mathcal{M}_0, \ldots, \mathcal{M}_p$ using an error metric.

# Testing and Training Splits & Error <a id='train-test'></a>

## Validation Set Approach - Training and Testing

The **validation set approach** splits data into *training* and *testing*. There are at least 2 problems with this approach:

1) The validation estimate of the test error rate can be highly variable.

2) If trained with fewer observations, the test error may be overestimating reality.

## Cross-Validation <a id='cv'></a>

## Leave One Out Cross Validation (LOOCV)

We can take an extreme approach and use all but 1 observation to build the model. This results in using $n-1$ observations to build each model. 

We calculate the test MSE ($MSE = \frac{1}{n} RSS$) for each model and then average the error,

$$
\text{CV}_{(n)} = \frac{1}{n} \sum^{n}_{i=1} \text{MSE}_{i} 
$$

## k-Fold Cross Validation 

An alternative to LOOCV is $k$-fold CV. This approach involves randomly dividing the set of observations into $k$ groups, or folds, of approximately equal size. Then, the error is given by,

$$
\text{CV}_{(k)} = \frac{1}{k} \sum^{k}_{i=1} \text{MSE}_i
$$

In practice, people typically use $k=3, 5,$ or $10$.
