# Local Stability Analysis <a id='stability'></a>

Recall that the ODEs describing the Lotka-Volterra model,

$$
\begin{align}
\dot{x} &=  \alpha x - \beta xy\\
\dot{y} &=  \delta xy - \gamma y
\end{align}
$$

The Lotka-Volterra model is an example of a nonlinear autonomous system of the form 

$$
\begin{align*}
\dot{x} &= f(x,y)\\
\dot{y} &= g(x,y)
\end{align*}
$$

or equivalently

$$
\dot{\mathbf{x}} = \mathbf{f(x)},
$$

where $\dot{\mathbf{x}} = [\dot{x}, \dot{y}]^T$ and $\mathbf{f(x)} = [f(x,y), g(x,y)]^T$

When we talk about stability, we mean finding the answer to this question:

**For equilibrium points, what is the stability of such steady states? If the point is  perturbed from its steady state, does it return to the equilibrium point or move away?**

To do this, we are going to calculate the Jacobian matrix $J$, which is the matrix of partial derivatives of the right-hand side of the ODE system, i.e., 

$$
J = \begin{bmatrix} \frac{\partial f}{\partial x} & \frac{\partial f}{\partial y}\\
 \frac{\partial g}{\partial x} & \frac{\partial g}{\partial y}\\
\end{bmatrix}.
$$

After calculating $J$, we evaluate it at the equilibrium points (also called fixed points) and calculate the eigenvalues at those points. Why?

Suppose that $\mathbf{x^∗}$ is an equilibrium point. By definition, $\mathbf{f(x^∗)} = 0.$ We can expand the right-hand side at this point, using multivariate Taylor series, as

$$
\begin{align}
\mathbf{\dot{x}} &= \mathbf{f(x^*)} + \left.\mathbf{\frac{\partial f}{\partial x}}\right|_{\mathbf{x^*}} \mathbf{(x - x^*)} + \ldots\\
    &= \left.\mathbf{\frac{\partial f}{\partial x}}\right|_{\mathbf{x^*}} \mathbf{(x - x^*)} + \ldots
\end{align}
$$

If we define $\mathbf{\delta x = x - x^*}$, taking the derivative of both sides (with respect to time), we get $\mathbf{ \dot{\delta x} = \dot{x}}.$ Then, near an equilibrium point (assuming $\delta$ is a small perturbation from the equilibrium point), we have

$$
\mathbf{\dot{\delta x} = J^* \delta x},
$$

where $J^*$ is the first partial derivative evaluated at the equilibrium point. This is just a **linear differential equation** and we can apply linear differential equation theory to classify the behavior of the solutions!

Since we are approximating this nonlinear system with a linear one **locally**, this analysis is known as **linear stability analysis**.

**Definition.** Let $p$ be a fixed point of the two-dimensional system $\dot{x} = f(x)$, and denote by $\lambda_1, \lambda_2$ the two eigenvalues of $\left. J\right|_p$ The following classification of the fixed point $p$ is standard.

1)  $\lambda_1, \lambda_2$ are real and distinct.

* $\lambda_1 \cdot \lambda_2> 0$ (the eigenvalues have the same sign): $p$ is called a node or type I singularity; that is, every orbit tends to the origin in a definite direction as $t \rightarrow \infty$.

    * $\lambda_1, \lambda_2$ > 0 : $p$ is an unstable node,
    * $\lambda_1, \lambda_2 < 0 $: $p$ is a stable node.

* $\lambda_1 \cdot \lambda_2$ < 0 (the eigenvalues have different signs): $p$ is an unstable fixed point, or a saddle point singularity.

2)  $\lambda_1, \lambda_2$ are complex, i.e., $\lambda_{1,2} = \alpha \pm i \beta, \beta \neq 0$.

* $\alpha \neq 0$ : $p$ is a spiral, or a focus, that is, the solutions approach the origin as $t \rightarrow \infty$, but not from a definite direction.
    * $\alpha < 0$: $p$ is a stable focus,
    * $\alpha > 0$: $p$ is an unstable focus.
    
* $\alpha = 0$: $p$ is a center, that means it is not stable in the usual sense, and we have to look at higher-order derivatives.

3)  $\lambda_1 = \lambda_2$.

* If there are two linearly independent eigenvectors, we have a stable singular node (these are simple straight lines through the origin).

* Calculate the eigenvalues of $J$ at the equilibrium points for the Lotka-Volterra model. Based on the information above, classify those points.

* Verify this numerically with numerical examples.

```python
def LK_rhs(Y, t, alpha, beta, delta, gamma):
# This function, LK_rhs, is the right-hand side of a simple version of the Lotka-Volterra model.

    # note: Y[0] refers to x and Y[1] refers to y
    # define dx/dt
    dx = alpha * Y[0] - beta * Y[0] * Y[1]
    # define dy/dt
    dy = delta * Y[0] * Y[1] - gamma * Y[1]
    # return both as a vector
    return [dx, dy]
```

```python
# define parameters
#alpha = 0.1
alpha = 0.5
beta = 0.02
gamma = 0.4
delta = 0.02
# define perturb as eps
eps = 0 #0.9
# define equilibrium points
eq1 = np.array([0,0])
eq2 = np.array([gamma/delta, alpha/beta])
# define initial condition
ic_eq1 = eq1 + eps
ic_eq2 = eq2 + eps
```

```python
# define an initial time
t0 = 0
# define an ending time
tf = 200
# define a linearly spaced grid to solve the system of equations (from t0 to tf)
t_out = np.linspace(t0, tf, num = 10000)
# define the parameters for the LK ODE system
param_vals = alpha, beta, delta, gamma
# solve numerically
sol_out1 = odeint(LK_rhs, ic_eq1, t_out, param_vals)
sol_out2 = odeint(LK_rhs, ic_eq1, t_out, param_vals)
```

```python
fig2 = go.Figure()
fig2.add_trace(go.Scatter(x=sol_out1[:,0], y=sol_out1[:,1], line_shape='spline'))
fig2.add_trace(go.Scatter(x=sol_out2[:,0], y=sol_out2[:,1], line_shape='spline'))


# Edit the layout
fig2.update_layout(title='Phase Plot',
                   xaxis_title='x(t)',
                   yaxis_title='y(t)')
fig2.show()
```
