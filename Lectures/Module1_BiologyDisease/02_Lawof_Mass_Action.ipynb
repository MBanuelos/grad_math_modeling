# Law of Mass Action <a id='mass-action'></a>

Biochemical reactions (and their applications) take into account **reactants** and **products**

A *basic synthesis reaction*: $$ A + B \rightarrow C $$
An *equilibrium reaction*: $$ A + B \rightleftharpoons C$$

In some cases, we'll have a *conservation of mass* - the mass of the products has to equal that of the reactants (closed system)

**Example:** 
Let us consider $$ A + B \overset{\lambda_{+}}\rightarrow C $$

The **law of mass action** states that the change of concentration in the product is proportional to the rates $\times$ the concentration of the reactants. Meaning, letting $[C]$ represent the concentration of $C$, we have

$$
\frac{d [C]}{dt} = \lambda_{+}[A][B]
$$

Given a family of chemical reactions, otherwise known as a chemical reaction network, the law of mass action provides an easy way to translate the reactions into a differential equation. Simply stated, consider the $n$ reactants $X_1, X_2, \ldots, X_n$ and $m$ reactions

$$
\sum_j^{n} a_{ij} X_j \underset{\lambda_i}{\rightarrow} \sum_j^{n} b_{ij} X_j 
$$

for $i = 1, 2, \ldots, m$ with rate $\lambda_i$. 

Then the concentration of reactant $X_k$, denoted $c_k$, satisfies

$$
\frac{d c_{k}}{dt} = \sum_i^{m} \lambda_i (b_{ik} - a_{ik}) \prod_j^{n} c_j^{a_{ij}}
$$

This is pretty reasonable assumption. A reaction like $ X + Y \rightarrow \ldots$ should intuitively occur both in proportion to the abundance of $X$ as well as the abundance of $Y$. It turns out that this assumption is good enough to power a good portion of the mathematical biology modeling literature.

If we again consider $$ A + B \underset{\lambda_{-}}{\stackrel{\lambda_{+}}{\rightleftharpoons}} C,$$ we can apply the previous equation as well.

# Case Study: Prion Disease <a id='prions'></a>

* Early research on scrapie, a degenerative disease affecting sheep, puzzled biologists.

* Transmission of information without going through nucleic acid was considered a fundamental violation of the *Central Dogma of Molecular Biology*

* Griffith (1967) suggested scrapie could be caused by a protein -- confirmed by Prusiner (1982) as “proteinaceous infectious particle,” or **prion for short**.

## Assumptions
* A prion aggregate consists of some number of constituent protein monomers, arranged linearly into a fiber. Thus, an aggregate has a “size” associated with it, which may analogously be considered its mass. 

* Aggregates increase in mass by incorporating a monomer into either end of the fiber, and they may spontaneously fragment into two smaller fibers (presumably due to the presence of cellular repair machinery). 

* We assume for simplicity that all possible fragmentations are equally likely. If any of the resulting fibers are smaller than some minimum size, they are assumed to immediately disassociate into their constituent monomers (hence the name “nucleated polymerization”).

## Prion Reactions

Let us denote the rates of monomer incorporation and fiber fragmentation as $\beta$ and $\gamma$, respectively. 

We will use the variable $n$ to denote the minimum size and additionally include protein synthesis (rate $\alpha$) and cell division (rate $\mu$) in our model. 

We first write the chemical reaction network summarizing the proposed dynamics, with reactants $X_1$ denoting protein monomers and $X_i$ $( i \ge n )$ prion aggregates of size $i$. We have

$$
\emptyset \underset{\mu}{\stackrel{\alpha}{\rightleftharpoons}} X_1,
$$

and then for each $i \ge n$, 

$$
X_1 + X_i \overset{2 \beta}{\rightarrow} X_{i+1}.
$$

For example,
$$
X_1 + X_4 \overset{2 \beta}{\rightarrow} X_5
$$
