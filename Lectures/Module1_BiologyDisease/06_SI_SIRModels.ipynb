## SI epidemiological models <a id='si-model'></a>

Susceptible-Infected **SI**, Susceptible-Infected-Recovered **SIR** epidemiological models (and the variants) are examples of **compartment models**, meaning you divide the population into different compartments and there is some interaction between the defined compartments.

These models are generally attributed to <a href="khttps://royalsocietypublishing.org/doi/10.1098/rspa.1927.0118"> William O. Kermack (Scottish biochemist) and Anderson G. McKendrick (Scottish physicist and epidemiologist) </a>. In this 1927 paper, they mention some common beliefs as to why an epidemic would stop:

* There are no more susceptible individuals
* The virus is not as potent as before

They suggested that it would actually depend on the relationship of population density, infection, recovery, and death rates.

These models can also be considered from a chemical reaction perspective.

**Q:** What are we trying to model and what does success mean?

### Assumptions for the SI model

* Individuals are born with no immunity (susceptible). 
* Once infected, individuals stay infected and infectious throughout their life, and remain in contact with the susceptible population. 
* The rate at which susceptible interact with infected and then become infected is $\beta$
* The population stays constant (no birth or death)

**These models have been used to study <a href="https://www.cdc.gov/cmv/overview.html"> cytomegalovirus (CMV)</a>, herpes, HIV, etc.**

$$
\begin{align}
S' &= - \beta S I\\
I' &= \beta S I
\end{align}
$$

# SIR epidemiological models <a id='sir-model'></a>

### Assumptions for the SIR model

Write a set of reactions that take all of these assumptions into account.


1) We consider just three classes of individuals: those who do not have the disease but are susceptible to it, those those who have the disease and are infective, and those who have recovered from the disease and can no longer be infected. Each individual is in one of these 3 classes. (In this model, no one is immune).

2) The rate at which infected recover is denoted by $\gamma$. The rate at which susceptible interact with infected and then become infected is $\beta$. 

3) The size of the population, $N$, remains fixed.

**These models have been used to study <a href="https://core.ac.uk/download/pdf/78556409.pdf"> influenza, Dengue fever, SARS, social networks, etc.</a>**

![sir model](https://upload.wikimedia.org/wikipedia/commons/9/93/SIR_Flow_Diagram.svg)

This leads to the following, classic SIR system

$$
\begin{align*}
S' &= -\beta SI \\
I' &= \beta SI - \gamma I \\
R' &= \gamma I
\end{align*}
$$
